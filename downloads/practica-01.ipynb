```python
import unittest
import sys

class TestSumar(unittest.TestCase):
    def test_positivos(self):
        self.assertEqual(sumar(2, 3), 5)
    def test_negativos(self):
        self.assertEqual(sumar(-2, -3), -5)
    def test_cero(self):
        self.assertEqual(sumar(0, 0), 0)
        self.assertEqual(sumar(4, 0), 4)
    def test_signos_distintos(self):
        self.assertEqual(sumar(-2, 3), 1)

resultado = unittest.TextTestRunner(stream=sys.stdout, verbosity=2).run(
    unittest.defaultTestLoader.loadTestsFromTestCase(TestSumar)
)
print("Todas las pruebas superadas" if resultado.wasSuccessful()
      else "Revisa tu función y vuelve a ejecutar las dos celdas")
```
